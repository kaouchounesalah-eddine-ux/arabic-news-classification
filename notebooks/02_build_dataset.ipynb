```python
import pandas as pd
df = pd.read_csv("../donnees/Le360.com.csv")
df
```

```python
category_mapping = {
    "Business": "اقتصاد",
    "Culture": "ثقافة",
    "International": "دولي",
    "Sport": "رياضة",
    "Policy": "سياسة",
    "Society": "مجتمع",
}
```

Keep only the six PFA categories

```python
df = df[df["Category"].isin(category_mapping)]
df
```


Translate categories into the PFA labels

```python
df["Category"] = df["Category"].map(category_mapping)
df
```

Categories:

```python
df["Category"].value_counts()
```

Select 1000 articles from each category:

```python
samples = []

for category in category_mapping.values():

    category_articles = df[df["Category"] == category]

    sample = category_articles.sample(
        n=3500,
        random_state=42
    )

    samples.append(sample)

df = pd.concat(samples, ignore_index=True)
df
```

Final category distribution and number of articles:

```python
df["Category"].value_counts()
# len(df)
```

Rename the columns

```python
df = df.rename(columns={
    "Title":"title",
    "Body":"body",
    "Category":'category'
})
df
```

Create the id

```python
df["id"]=[
    f"art-{i:04d}"
    for i in range(1,len(df)+1)
]
df
```

```python
df["categories"]=df["category"].apply(lambda x:[x])

```

```python
import random
from datetime import date, timedelta

random.seed(42)

start_date = date(2024, 1, 1)
end_date = date(2026, 3, 31)

days_range = (end_date - start_date).days

df["publishedAt"] = [
    (start_date + timedelta(days=random.randint(0, days_range))).isoformat()
    for _ in range(len(df))
]
df=df[["id","title","body","categories","publishedAt"]]
df
```

Convert the DataFrame to JSON

```python
data = df.to_dict(orient="records")

import json

with open(
    "../donnees/articles.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        data,
        f,
        ensure_ascii=False,
        indent=2
    )
df
```
